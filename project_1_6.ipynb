# Identifying Ideal Customers (RFM Model)

Recency can be determined by taking the max of the sale_date for the sales associated with each customer. To do this, the query would select the customer_id from the customer table. It would also select the max function applied to the sale_date from the sales table. Using the max function would return the most recent date of a customer's sale. However, this is not necessarily the easiest way to score customers. A recency score can be determined by using two extract statements: one to obtain the week in the year the sale was made and one to obtain the day of the month of the sale. These two values can then be multiplied together to determine a recency score. Higher values would indicate more recent customers. The sales table would be joined to the customers table on the customer_id. The query would have to have a group by clause including the customer_id.

Frequency can be calculated by counting the number of sales made in 2020. Since only one order can be placed per customer per day, using a count function on a customer's sale_id's yields the number of times they visited AGM that year. This could be accomplished by selecting the customer_id from the customer table. It would also count the sale_id from the sale's table. The customer table would be joined to the sales table on the customer_id. The query would have to be grouped by the customer_id.

Monetary value can be determined by taking the mean of the total_amount for each customers sales. Using the average instead of a raw summation of the total_amount helps account for large, one-off purchases. This query would select the customer_id from the customer's table and the average of the total_ammount from the sales table. The customer's and sales tables would be joined on the customer_id. The query would have to be grouped by the customer_id.

The results of the three queries above will be combined to create a final customer score. Because higher recency, frequency, and monetary value scores all indicate better customers, the final synthesis score will be the multiplication of all three. This will result in a fairly large number, so to scale the score to a more comprehendable value, it will be divided by 1000. The three subqueries can be joined on the customer_id. The final query can then order by the synthesis score to determine the best customers.
